# Using sklearn's Linear Regression

```python
from sklearn.model_selection import train_test_split # Import train_test_split for splitting data into training and testing sets
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 2) # Split the dataset into training and testing sets
```

```python
print(X_train.shape) # Display the shape of the training features
print(X_test.shape) # Display the shape of the testing features
```

```python
from sklearn.linear_model import LinearRegression # Import LinearRegression model from sklearn
```

```python
lr = LinearRegression() # Create an instance of the Linear Regression model
```

```python
lr.fit(X_train, y_train) # Train the Linear Regression model using the training data
```

```python
y_pred = lr.predict(X_test) # Make predictions on the test data
```

```python
from sklearn.metrics import r2_score # Import r2_score for evaluating model performance
```

```python
print("R2 Score: ", r2_score(y_test, y_pred)) # Calculate and print the R2 score for the model
```

```python
lr.coef_ # Display the coefficients (slopes) of the linear regression model
```

```python
lr.intercept_ # Display the intercept of the linear regression model
```

# Making our own Linear Regression Class

```python
class MeraLR:
  # Custom Linear Regression class from scratch
  def __init__(self):
    self.coef_ = None # Initialize coefficients to None
    self.intercept_ = None # Initialize intercept to None

  def fit(self, X_train, y_train):
    # Add a column of ones to X_train for the intercept term
    X_train = np.insert(X_train, 0, 1, axis = 1)

    # Calculate the coefficients (betas) using the normal equation: (X_T * X)^-1 * X_T * y
    betas = np.linalg.inv(np.dot(X_train.T, X_train)).dot(X_train.T).dot(y_train)
    self.intercept_ = betas[0] # The first element is the intercept
    self.coef_ = betas[1:] # The remaining elements are the coefficients for features

  def predict(self, X_test):
    # Make predictions using the formula: y_pred = X_test * coef_ + intercept_
    y_pred = np.dot(X_test, self.coef_) + self.intercept_
    return y_pred
```

```python
reg = MeraLR() # Create an instance of our custom Linear Regression model
```

```python
reg.fit(X_train, y_train) # Train our custom Linear Regression model
```

```python
y_pred = reg.predict(X_test) # Make predictions on the test data using our custom model
```

```python
print("R2 Score: ", r2_score(y_test, y_pred)) # Calculate and print the R2 score for our custom model
```
