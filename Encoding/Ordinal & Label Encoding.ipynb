```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/customer.csv')
```

```python
df.sample(5)
```

```python
df = df.iloc[:,2:] # Select all rows and columns starting from the third column (index 2) onwards.
# This effectively drops the 'age' and 'gender' columns, keeping 'review', 'education', and 'purchased'.
```

```python
from sklearn.model_selection import train_test_split
# Split the DataFrame into features (X) and target (y) for training and testing.
# X will contain 'review' and 'education' columns.
# y will contain the 'purchased' column.
# test_size=0.2 means 20% of the data will be used for testing, and 80% for training.
X_train, X_test, y_train, y_test = train_test_split(df[['review', 'education']], df['purchased'], test_size=0.2)
```

```python
# Import OrdinalEncoder for converting categorical features to numerical (ordinal) data
from sklearn.preprocessing import OrdinalEncoder
```

```python
X_train
```

```python
# Initialize OrdinalEncoder with the specified categories for 'review' and 'education' columns.
# This defines the order for encoding: 'Poor' < 'Average' < 'Good' and 'School' < 'UG' < 'PG'.
oe = OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])
```

```python
# Apply the OrdinalEncoder to the training data (X_train).
# fit_transform learns the categories from X_train and then transforms it into numerical format.
# The previous error occurred because X_train was empty. With the fix in the train_test_split, this should now work.
X_train = oe.fit_transform(X_train)
```
