```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/placement.csv')
```

```python
# Check for any missing values in each column of the DataFrame.
df.isnull().sum()
```

```python
# Create a figure with two subplots side-by-side to visualize the distribution of 'cgpa' and 'placement_exam_marks'.
plt.figure(figsize=(16,5))

# First subplot: Distribution plot for 'cgpa' column.
# Using 'histplot' as 'distplot' is deprecated.
plt.subplot(1,2,1)
sns.histplot(df['cgpa'], kde=True)
plt.title('Distribution of CGPA')
plt.xlabel('CGPA')
plt.ylabel('Density')

# Second subplot: Distribution plot for 'placement_exam_marks' column.
# Using 'histplot' as 'distplot' is deprecated.
plt.subplot(1,2,2)
sns.histplot(df['placement_exam_marks'], kde=True)
plt.title('Distribution of Placement Exam Marks')
plt.xlabel('Placement Exam Marks')
plt.ylabel('Density')

# Display the plots.
plt.tight_layout()
plt.show()
```

```python
# Calculate and print the mean value of 'cgpa'.
print('Mean value of cgpa', df['cgpa'].mean())
# Calculate and print the standard deviation of 'cgpa'.
print('Std value of cgpa', df['cgpa'].std())
# Calculate and print the minimum value of 'cgpa'.
print('Min value of cgpa', df['cgpa'].min())
# Calculate and print the maximum value of 'cgpa'.
print('Max value of cgpa', df['cgpa'].max())
```

```python
# Finding the upper and lower boundary values for 'cgpa' based on the 3-sigma rule (mean +/- 3 * standard deviation).
# Print the highest allowed value.
print('Highest allowed', df['cgpa'].mean() + 3*df['cgpa'].std())
# Print the lowest allowed value.
print('Lowest allowed', df['cgpa'].mean() - 3*df['cgpa'].std())
```

```python
# Finding the outliers in the 'cgpa' column.
# Outliers are defined as values above 8.80 or below 5.11 (based on the 3-sigma rule).
df[(df['cgpa'] > 8.80) | (df['cgpa'] < 5.11)]
```

# Trimming: Removing Outliers

```python
# Trimming the DataFrame to remove outliers in the 'cgpa' column.
# Create a new DataFrame 'new_df' containing only rows where 'cgpa' is between 5.11 and 8.80 (exclusive).
new_df = df[(df['cgpa'] < 8.80) & (df['cgpa'] > 5.11)]
```
