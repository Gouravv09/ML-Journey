```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/placement (1).csv')
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'], df['package']) # Create a scatter plot of CGPA vs. Package
plt.xlabel('CGPA') # Label the x-axis
plt.ylabel('Package in Lpa') # Label the y-axis
plt.title('Student Record') # Set the plot title
```

```python
X = df.iloc[:,0:1] # Select 'cgpa' as the feature (X)
y = df.iloc[:,-1] # Select 'package' as the target (y)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2, random_state = 2) # Split data into training and testing sets
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression() # Initialize the Linear Regression model
```

```python
lr.fit(X_train, y_train) # Train the Linear Regression model
```

```python
X_test # Display the test features
```

```python
y_test # Display the test target values
```

```python
lr.predict(X_test.iloc[0].values.reshape(1,1)) # Predict package for the first test CGPA
```

```python
lr.predict(X_test.iloc[1].values.reshape(1,1)) # Predict package for the second test CGPA
```

```python
lr.predict(X_test.iloc[2].values.reshape(1,1)) # Predict package for the third test CGPA
```

```python
plt.scatter(df['cgpa'], df['package']) # Plot the original scatter data
plt.plot(X_train, lr.predict(X_train), color = 'red') # Plot the regression line
plt.xlabel('CGPA') # Label the x-axis
plt.ylabel('Package in Lpa') # Label the y-axis
plt.title('Student Record') # Set the plot title
```

```python
m = lr.coef_ # Get the coefficient (slope) of the regression line
m
```

```python
b = lr.intercept_ # Get the intercept of the regression line
b
```
