```python
class MeraLR:

    def __init__(self):
        self.m = None # Initialize slope (m)
        self.b = None # Initialize intercept (b)

    def fit(self,X_train,y_train):

        num = 0 # Initialize numerator for slope calculation
        den = 0 # Initialize denominator for slope calculation

        for i in range(X_train.shape[0]):

            # Calculate numerator for slope (m)
            num = num + ((X_train[i] - X_train.mean())*(y_train[i] - y_train.mean()))
            # Calculate denominator for slope (m)
            den = den + ((X_train[i] - X_train.mean())*(X_train[i] - X_train.mean()))

        self.m = num/den # Calculate the slope (m)
        self.b = y_train.mean() - (self.m * X_train.mean()) # Calculate the intercept (b)
        print(self.m) # Print the calculated slope
        print(self.b) # Print the calculated intercept

    def predict(self,X_test):

        print(X_test) # Print the test data point

        return self.m * X_test + self.b # Return the predicted value
```
